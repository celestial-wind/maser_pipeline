```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import windows
import cmocean

# Set plot style for publication quality
plt.style.use('default')
plt.rcParams.update({
    'font.size': 12,
    'axes.labelsize': 12,
    'xtick.labelsize': 11,
    'ytick.labelsize': 11,
    'legend.fontsize': 11,
    'font.family': 'serif',
})

# --- Helper Functions from the Simulation ---

# Constants for CHIME
N_CHANNELS = 1024
BW_START = 400.0  # MHz
BW_END = 800.0    # MHz
NATIVE_CHANNEL_WIDTH = (BW_END - BW_START) / N_CHANNELS

def create_chime_pfb_window_direct(N, ntap=4):
    """Direct implementation for CHIME PFB power response kernel."""
    k = np.arange(N) - N // 2
    sinc_squared = np.sinc(ntap * k / N) ** 2
    hamming_window = windows.hamming(N) ** 2
    return sinc_squared * hamming_window

def apply_pfb_smoothing(spectrum: np.ndarray) -> np.ndarray:
    """Simulates PFB spectral leakage by convolving with a sinc^2-Hamming kernel."""
    kernel = create_chime_pfb_window_direct(len(spectrum), ntap=4)
    kernel /= np.sum(kernel)  # Normalize to conserve flux
    return np.convolve(spectrum, kernel, mode='same')

def rebin_spectrum(spec_hr: np.ndarray, freq_hr: np.ndarray, freqs_native: np.ndarray) -> np.ndarray:
    """Bins a high-resolution spectrum down to a native, lower-resolution grid."""
    bin_edges = np.zeros(len(freqs_native) + 1)
    bin_edges[1:-1] = (freqs_native[:-1] + freqs_native[1:]) / 2
    bin_edges[0] = freqs_native[0] - NATIVE_CHANNEL_WIDTH / 2
    bin_edges[-1] = freqs_native[-1] + NATIVE_CHANNEL_WIDTH / 2

    digitized = np.digitize(freq_hr, bin_edges)
    valid_mask = (digitized > 0) & (digitized <= len(freqs_native))
    
    min_len = len(freqs_native) + 1
    bin_sums = np.bincount(digitized[valid_mask], weights=spec_hr[valid_mask], minlength=min_len)
    bin_counts = np.bincount(digitized[valid_mask], minlength=min_len)

    binned_spec = np.zeros_like(freqs_native, dtype=float)
    counts_for_bins = bin_counts[1:]
    sums_for_bins = bin_sums[1:]
    
    non_empty_bins = counts_for_bins > 0
    binned_spec[non_empty_bins] = sums_for_bins[non_empty_bins] / counts_for_bins[non_empty_bins]
    return binned_spec

# --- 1. Generate Mock Data ---

freq_hr = np.linspace(599, 601, 8192)
line_center = 600.0
line_width = 0.15
spec_hr = np.exp(-0.5 * ((freq_hr - line_center) / line_width) ** 2)
freqs_native = np.linspace(BW_START, BW_END - NATIVE_CHANNEL_WIDTH, N_CHANNELS)

# --- 2. Run the Simulation ---

spec_smooth = apply_pfb_smoothing(spec_hr)
spec_binned = rebin_spectrum(spec_smooth, freq_hr, freqs_native)

# --- 3. Create the Plot ---

fig, ax = plt.subplots(figsize=(9, 6), dpi=150)
colors = cmocean.cm.haline([0.2, 0.55, 0.85])

# Plot spectral lines on the primary y-axis
ax.plot(freq_hr, spec_hr, color=colors[0], ls='--', lw=1.5, label='High-Res Model')
ax.plot(freq_hr, spec_smooth, color=colors[1], lw=2.0, label='PFB-Smoothed Model')
ax.step(freqs_native, spec_binned, where='mid', color=colors[2], lw=2.5, label='Final Binned Spectrum')

# Plot PFB filter response on the secondary y-axis
ax2 = ax.twinx()
pfb_kernel = create_chime_pfb_window_direct(len(freq_hr), ntap=8)
pfb_kernel /= np.max(pfb_kernel)
kernel_freq_axis = freq_hr - np.mean(freq_hr) + line_center
ax2.plot(kernel_freq_axis, pfb_kernel, color='k', lw=1.5, alpha=0.3, label='PFB Power Response')

# --- Annotations and Visual Guides ---
target_channel_idx = np.argmin(np.abs(freqs_native - line_center))
ch_center = freqs_native[target_channel_idx]
ch_width = NATIVE_CHANNEL_WIDTH
ax.axvspan(ch_center - ch_width/2, ch_center + ch_width/2, color='grey', alpha=0.15, label='Central Channel Bin')

# !! NEW: Add vertical lines for adjacent channel centers !!
adj_ch_minus_1_freq = freqs_native[target_channel_idx - 1]
adj_ch_plus_1_freq = freqs_native[target_channel_idx + 1]
ax.axvline(adj_ch_minus_1_freq, color='indianred', ls=':', lw=1.5, alpha=0.9, label='Adjacent Channel Center')
ax.axvline(adj_ch_plus_1_freq, color='indianred', ls=':', lw=1.5, alpha=0.9)


# --- Final Formatting ---
ax.set_xlabel('Frequency [MHz]')
ax.set_ylabel('Relative Flux Density')
# ax.set_title('Simulation of CHIME PFB Response and Channel Overlap', pad=15)
ax.set_xlim(599.2, 600.8)
ax.set_ylim(-0.05, 1.1)
ax.tick_params(axis='y', labelcolor='k')

ax2.set_ylabel('PFB Power Response [log scale]', color='k')
ax2.set_yscale('log')
ax2.set_ylim(1e-6, 2)
ax2.tick_params(axis='y', labelcolor='k')

ax.text(0.97, 0.97, 'PFB Model: 4-tap', transform=ax.transAxes, ha='right', va='top',
        bbox=dict(boxstyle="round,pad=0.3", fc="w", ec="k", lw=0.5, alpha=0.6))

# Create a combined legend for both axes
lines, labels = ax.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax.legend(lines + lines2, labels + labels2, loc='upper left', fontsize=9)

ax.grid(True, which='major', linestyle='--', linewidth=0.5, alpha=0.7)
fig.tight_layout()
fig.savefig('PFB_example.jpeg', dpi=150)
plt.show()
```

Of course. The original equation combines the distinct steps of smoothing and rebinning into one expression and incorrectly defines the filter kernel.

Here is a revised set of equations that more accurately reflects the two-step process used in the simulation code, followed by an explanation of the changes.

### Corrected Equations

The process is best described in three parts: the spectral smoothing, the kernel definition, and the final rebinning.

1.  **Spectral Smoothing (Convolution):** The high-resolution input spectrum, $S_{\text{hr}}$, is first convolved with the normalized PFB power response kernel, $K_{\text{norm}}$.
    \begin{equation}
    S_{\text{smooth}}(\nu) = (S_{\text{hr}} * K_{\text{norm}})(\nu)
    \end{equation}

2.  **PFB Power Response Kernel:** The kernel is defined directly in the frequency domain as the product of a squared-sinc function and a windowing function.
    \begin{equation}
    K(\nu'_{\text{norm}}) = \underbrace{\left[ \text{sinc}\left(N_{\text{tap}} \cdot \nu'_{\text{norm}}\right) \right]^2}_{\text{Ideal PFB Power Response}} \cdot \underbrace{W_{\text{Hamming}}(\nu'_{\text{norm}})}_{\text{Sidelobe Suppression}}
    \end{equation}
    where $N_{\text{tap}}$ is the number of PFB taps (e.g., 4) and $\nu'_{\text{norm}}$ is the frequency offset normalized to the bandwidth of the kernel. This kernel is then normalized to have an integral of one to conserve flux.

3.  **Rebinning (Integration):** The smoothed spectrum is then averaged over each discrete, native frequency channel.
    \begin{equation}
    S_{\text{binned}}[k] = \frac{1}{\Delta \nu_{\text{native}}} \int_{\nu_k - \Delta\nu_{\text{native}}/2}^{\nu_k + \Delta\nu_{\text{native}}/2} S_{\text{smooth}}(\nu) \,d\nu
    \end{equation}
    where $S_{\text{binned}}[k]$ is the final value for the k-th channel centered at frequency $\nu_k$, and $\Delta\nu_{\text{native}}$ is the width of a native channel.

### Explanation of Changes

1.  **Separation of Steps:** The original equation conflated the smoothing and rebinning. The corrected version separates the physics into two distinct operations: a **convolution** (Eq. 1) followed by an **integration** (Eq. 3). This directly mirrors the code's logic of `apply_pfb_smoothing()` followed by `rebin_spectrum()`.

2.  **Correct Kernel Definition:** The original equation incorrectly defined the PFB response as the Fourier transform of a time-domain filter. Our simulation defines the kernel **directly in the frequency domain** (Eq. 2), which is what `create_chime_pfb_window_direct` does. It correctly shows the power response as a `sinc` **squared** function multiplied by a Hamming window.

3.  **Clearer Rebinning:** The corrected rebinning equation (Eq. 3) clearly shows that the final value is the **average** of the **smoothed** spectrum over the boundaries of a single native channel. This is precisely what the `np.bincount(sums) / np.bincount(counts)` operation in `rebin_spectrum` achieves.

Yes, here is a single, two-line mathematical expression that condenses the process and corrects the original formulation.

\begin{equation}
\begin{aligned}
S_{k} &= \frac{1}{\Delta\nu_{\text{native}}} \int_{\text{bin } k} (S_{\text{hr}} * K)(\nu) \,d\nu \\
&\qquad K(\nu'_{\text{norm}}) = \left[ \text{sinc}\left(N_{\text{tap}} \cdot \nu'_{\text{norm}}\right) \right]^2 \cdot W_{\text{Hamming}}(\nu'_{\text{norm}})
\end{aligned}
\end{equation}

This expression states that the final value in a channel, $S_k$, is the average value of the smoothed spectrum over that channel's bin. The smoothed spectrum itself is the result of convolving ($*$) the high-resolution input spectrum ($S_{\text{hr}}$) with the PFB power response kernel ($K$), which is defined in the second line.

Of course. Here is the discrete version of the expression, which more directly reflects the array-based operations in the Python code.

\begin{equation}
\begin{aligned}
S_{k} &= \frac{1}{|I_k|} \sum_{i \in I_k} \left( \sum_{m} S_{\text{hr}}[i-m] \cdot K[m] \right) \\
&\qquad K[m] = \left[ \text{sinc}\left(\frac{N_{\text{tap}} \cdot m}{N}\right) \right]^2 \cdot W_{\text{Hamming}}[m]
\end{aligned}
\end{equation}

***

### Explanation of Terms

This discrete form precisely maps to the simulation's logic:

* $S_k$ is the final value for the **k-th** native channel.
* The inner sum, $\sum_{m} S_{\text{hr}}[i-m] \cdot K[m]$, is the **discrete convolution** (`np.convolve`) that performs the PFB smoothing on the high-resolution spectrum, $S_{\text{hr}}$.
* $I_k$ is the set of high-resolution indices **i** that fall within the frequency boundaries of the **k-th** native channel.
* The outer sum, $\sum_{i \in I_k}$, represents the **summation** of all smoothed data points within the channel bin, which `np.bincount` calculates.
* $\frac{1}{|I_k|}$ is the **averaging** factor, where $|I_k|$ is the count of high-resolution points in the bin.
* $K[m]$ is the discrete **PFB kernel array**, where **m** is the index offset from the center, **N** is the total number of points in the kernel, and $W_{\text{Hamming}}$ is the discrete Hamming window.

Of course. The functions `np.convolve`, `np.digitize`, and `np.bincount` are critical optimizations that leverage NumPy's high-performance, vectorized capabilities. They replace manual Python loops, which are significantly slower.

These functions are written in low-level, compiled languages (like C or Fortran) and are designed to operate on entire arrays at once, avoiding the overhead of interpreting Python code for each element in a loop.

---
### ## `np.convolve`

* **Its Role** ⚙️: In `apply_pfb_smoothing`, this function performs the convolution of the high-resolution spectrum with the PFB kernel. This is the core mathematical operation for simulating the filter's smoothing effect.

* **Why It's Faster**: For large arrays, `np.convolve` automatically uses an extremely efficient algorithm based on the **Fast Fourier Transform (FFT)**. The convolution theorem states that convolution in the time/frequency domain is equivalent to simple multiplication in the frequency/time domain. The FFT algorithm has a time complexity of roughly $O(N \log N)$, which is vastly superior to a manual implementation.

* **The Slower Alternative**: Without `np.convolve`, you would have to write the convolution as a set of nested `for` loops, manually sliding the kernel across the spectrum array, multiplying, and summing at each step. This has a time complexity of $O(N \cdot M)$ (where N and M are the lengths of the spectrum and kernel), which is much slower for the large arrays used in simulations.

---
### ## `np.digitize`

* **Its Role** 🗂️: In `rebin_spectrum`, `np.digitize` takes each frequency value from the high-resolution spectrum and quickly determines which native channel bin it belongs to.

* **Why It's Faster**: This function uses an efficient **binary search** (`np.searchsorted`) because the `bin_edges` array is sorted. A binary search has a time complexity of $O(\log K)$ for each point, leading to a total time of $O(N \log K)$ for N points and K bins.

* **The Slower Alternative**: The manual approach would be to loop through each of the `N` high-resolution frequencies. Inside that loop, you would have another loop that iterates through the `K` bin edges to find the correct one. This brute-force search would have a time complexity of $O(N \cdot K)$, which is significantly worse.

---
### ## `np.bincount`

* **Its Role** ➕: This is the workhorse of the `rebin_spectrum` function. After `np.digitize` assigns each data point to a bin index, `np.bincount` performs the actual summation and counting needed for the average. It is used twice: once without weights to count the number of high-res points in each bin, and once with `weights=spec_hr` to sum the flux values in each bin.

* **Why It's Faster**: `np.bincount` is arguably the most significant optimization here. It's a highly specialized and heavily optimized C function that performs this "group-by and sum" operation in a single pass with no Python-level looping. It is one of the fastest ways to perform this type of histogramming or binning task in Python.

* **The Slower Alternative**: To replace `np.bincount`, you would first have to initialize two arrays of zeros (e.g., `bin_sums` and `bin_counts`). Then, you would have to write a Python `for` loop to iterate through all the digitized indices and their corresponding spectral values, adding the value to `bin_sums` and incrementing the counter in `bin_counts`. This loop introduces massive overhead and is orders of magnitude slower than the single, optimized `np.bincount` calls.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import windows

def create_pfb_kernel(n_points_fft, n_channels=1024, ntap=4):
    """
    Generates a PFB power response kernel from a time-domain FIR filter design.

    Args:
        n_points_fft (int): The number of points for the output kernel FFT.
        n_channels (int): The number of PFB channels.
        ntap (int): The number of taps for the filter.

    Returns:
        tuple: Frequency axis (in bins) and the normalized power response kernel.
    """
    # 1. Create the FIR filter coefficients in the time domain
    M = ntap * n_channels
    n = np.arange(M) - M // 2
    sinc_coeffs = np.sinc(n / n_channels)
    hamming_window = windows.hamming(M)
    filter_coeffs = sinc_coeffs * hamming_window

    # 2. FFT to get the frequency response
    padded_coeffs = np.zeros(n_points_fft)
    # Center the filter coefficients in the padded array for phase correctness
    start_index = (n_points_fft - M) // 2
    padded_coeffs[start_index : start_index + M] = filter_coeffs
    
    freq_response = np.fft.fft(padded_coeffs)
    
    # 3. Get the power response and shift for plotting
    power_response = np.abs(freq_response)**2
    power_response_shifted = np.fft.fftshift(power_response)
    
    # 4. Create frequency axis in units of channels/bins
    freq_axis_bins = (np.arange(n_points_fft) - n_points_fft // 2) / (n_points_fft / n_channels)
    
    return freq_axis_bins, power_response_shifted / np.max(power_response_shifted)

# --- Generate and Plot ---
N_FFT = 16384 # High-resolution FFT for a smooth plot

# Generate both 4-tap and 8-tap kernels
freq_axis, kernel_4tap = create_pfb_kernel(N_FFT, ntap=4)
_, kernel_8tap = create_pfb_kernel(N_FFT, ntap=8)

# Convert to dB for clear comparison
power_db_4tap = 10 * np.log10(kernel_4tap)
power_db_8tap = 10 * np.log10(kernel_8tap)

# --- Plotting ---
plt.figure(figsize=(12, 8))
plt.plot(freq_axis, power_db_4tap, lw=2, label='4-Tap Sinc-Hamming (Your current model)')
plt.plot(freq_axis, power_db_8tap, lw=2, label='8-Tap Sinc-Hamming (Suggested model)', c='darkorange', ls='--')

plt.title('PFB Performance: 4 Taps vs. 8 Taps', fontsize=16)
plt.xlabel('Frequency Offset from Channel Center (bins)', fontsize=12)
plt.ylabel('Power (dB)', fontsize=12)
plt.xlim(-4, 4)
plt.ylim(-120, 5)
plt.grid(True, which='both', linestyle=':')
plt.axhline(y=-42, color='red', ls=':', lw=1.5, label='Hamming Window Sidelobe Limit (~ -42 dB)')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import windows
from scipy.interpolate import interp1d
import cmocean

# Set plot style for publication quality
plt.style.use('default')
plt.rcParams.update({
    'font.size': 12, 'axes.labelsize': 12, 'xtick.labelsize': 11,
    'ytick.labelsize': 11, 'legend.fontsize': 11, 'font.family': 'serif',
})

# --- Constants for CHIME ---
N_CHANNELS = 1024
BW_START = 400.0
BW_END = 800.0
FULL_BW = BW_END - BW_START
NATIVE_CHANNEL_WIDTH = FULL_BW / N_CHANNELS

# --- Helper Functions from the Simulation ---

# NEW: A robust function that generates the PFB response on any target frequency axis
def create_pfb_response(target_freq_mhz: np.ndarray, center_freq_mhz: float, ntap: int = 4) -> np.ndarray:
    """
    Accurately models the PFB power response and interpolates it onto a target frequency axis.

    Args:
        target_freq_mhz: The frequency grid (in MHz) for the output kernel.
        center_freq_mhz: The frequency (in MHz) where the PFB response should be centered.
        ntap: The number of PFB taps.

    Returns:
        The PFB power response kernel on the target frequency grid.
    """
    # 1. Design the FIR filter in the time domain (unchanged)
    M = ntap * N_CHANNELS
    time_idx = np.arange(M) - M // 2
    filter_coeffs = np.sinc(time_idx / N_CHANNELS) * windows.hamming(M)

    # 2. Generate a smooth, high-resolution model of the channel shape via FFT.
    # This FFT spans the *entire* 400 MHz CHIME bandwidth.
    N_fft_model = 32768  # Use a large FFT for a very smooth model
    power_response_model = np.abs(np.fft.fft(filter_coeffs, N_fft_model))**2
    power_response_model_shifted = np.fft.fftshift(power_response_model)

    # 3. Create the correct frequency axis for this model (in MHz).
    # The frequency step is determined by the full bandwidth and FFT length.
    model_freq_axis = np.fft.fftfreq(N_fft_model, d=1.0/FULL_BW)
    model_freq_axis_shifted = np.fft.fftshift(model_freq_axis)

    # 4. Interpolate the high-res model onto the user's specific frequency grid.
    target_freq_offsets = target_freq_mhz - center_freq_mhz
    interpolator = interp1d(model_freq_axis_shifted, power_response_model_shifted,
                            bounds_error=False, fill_value=0)
    interpolated_response = interpolator(target_freq_offsets)
    
    return interpolated_response

def apply_pfb_smoothing(spectrum: np.ndarray, freq_hr: np.ndarray, ntap: int = 4) -> np.ndarray:
    """Simulates PFB spectral leakage using the accurately scaled kernel."""
    # The kernel is now generated specifically for our frequency axis
    kernel = create_pfb_response(freq_hr, np.mean(freq_hr), ntap=ntap)
    kernel /= np.sum(kernel)  # Normalize for flux conservation
    return np.convolve(spectrum, kernel, mode='same')

# The rebin_spectrum function is unchanged
def rebin_spectrum(spec_hr: np.ndarray, freq_hr: np.ndarray, freqs_native: np.ndarray) -> np.ndarray:
    """Bins a high-resolution spectrum down to a native, lower-resolution grid."""
    bin_edges = np.zeros(len(freqs_native) + 1)
    bin_edges[1:-1] = (freqs_native[:-1] + freqs_native[1:]) / 2
    bin_edges[0] = freqs_native[0] - NATIVE_CHANNEL_WIDTH / 2
    bin_edges[-1] = freqs_native[-1] + NATIVE_CHANNEL_WIDTH / 2
    digitized = np.digitize(freq_hr, bin_edges)
    valid_mask = (digitized > 0) & (digitized <= len(freqs_native))
    min_len = len(freqs_native) + 1
    bin_sums = np.bincount(digitized[valid_mask], weights=spec_hr[valid_mask], minlength=min_len)
    bin_counts = np.bincount(digitized[valid_mask], minlength=min_len)
    binned_spec = np.zeros_like(freqs_native, dtype=float)
    counts_for_bins = bin_counts[1:]
    sums_for_bins = bin_sums[1:]
    non_empty_bins = counts_for_bins > 0
    binned_spec[non_empty_bins] = sums_for_bins[non_empty_bins] / counts_for_bins[non_empty_bins]
    return binned_spec

# --- 1. Generate Mock Data ---
freq_hr = np.linspace(599, 601, 8192)
line_center = 600.0
spec_hr = np.exp(-0.5 * ((freq_hr - line_center) / 0.15)**2)
freqs_native = np.linspace(BW_START, BW_END - NATIVE_CHANNEL_WIDTH, N_CHANNELS)

# --- 2. Run the Simulation ---
spec_smooth = apply_pfb_smoothing(spec_hr, freq_hr, ntap=4)
spec_binned = rebin_spectrum(spec_smooth, freq_hr, freqs_native)

# --- 3. Create the Plot ---
fig, ax = plt.subplots(figsize=(9, 6), dpi=150)
colors = cmocean.cm.haline([0.2, 0.55, 0.85])

ax.plot(freq_hr, spec_hr, color=colors[0], ls='--', lw=1.5, label='High-Res Model')
ax.plot(freq_hr, spec_smooth, color=colors[1], lw=2.0, label='PFB-Smoothed Model')
ax.step(freqs_native, spec_binned, where='mid', color=colors[2], lw=2.5, label='Final Binned Spectrum')

ax2 = ax.twinx()
# Generate the response for plotting (it's the same kernel used for smoothing)
pfb_response_for_plot = create_pfb_response(freq_hr, line_center, ntap=4)
pfb_response_db = 10 * np.log10(pfb_response_for_plot / np.max(pfb_response_for_plot))

# Plot the kernel against the high-res frequency axis
ax2.plot(freq_hr, pfb_response_for_plot/np.max(pfb_response_for_plot), color='k', lw=2.0, alpha=0.4, label='PFB Power Response (dB)')

# --- Annotations and Visual Guides ---
target_channel_idx = np.argmin(np.abs(freqs_native - line_center))
ch_center = freqs_native[target_channel_idx]
ch_width = NATIVE_CHANNEL_WIDTH
ax.axvspan(ch_center - ch_width/2, ch_center + ch_width/2, color='grey', alpha=0.15, label='Central Channel Bin')

# !! NEW: Add vertical lines for adjacent channel centers !!
adj_ch_minus_1_freq = freqs_native[target_channel_idx - 1]
adj_ch_plus_1_freq = freqs_native[target_channel_idx + 1]
ax.axvline(adj_ch_minus_1_freq, color='indianred', ls=':', lw=1.5, alpha=0.9, label='Adjacent Channel Center')
ax.axvline(adj_ch_plus_1_freq, color='indianred', ls=':', lw=1.5, alpha=0.9)


# --- Final Formatting ---
ax.set_xlabel('Frequency [MHz]')
ax.set_ylabel('Relative Flux Density')
ax.set_xlim(599.2, 600.8)
ax.set_ylim(-0.05, 1.1)
ax2.set_ylabel('PFB Power Response [dB]')
ax2.set_ylim(-100, 5)
lines, labels = ax.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax.legend(lines + lines2, labels + labels2, loc='upper left', fontsize=10)
ax.grid(True, which='major', linestyle='--', linewidth=0.5, alpha=0.7)
fig.tight_layout()
fig.savefig('PFB_example_corrected.jpeg', dpi=150)
plt.show()
```
