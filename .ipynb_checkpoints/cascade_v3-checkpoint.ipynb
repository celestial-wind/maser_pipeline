```python
import matplotlib.pyplot as plt
import numpy as np
import cmocean
from matplotlib.patches import FancyBboxPatch
from matplotlib.patches import Circle
import matplotlib.patches as mpatches

# Set up the figure
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 10))
fig.suptitle('Turbulent Energy Cascade: Kolmogorov vs Gravitomagnetic', 
             fontsize=20, fontweight='bold', y=0.95)

# Color scheme using cmocean thermal
cmap = cmocean.cm.thermal
colors = [cmap(0.2), cmap(0.4), cmap(0.6), cmap(0.8)]
bg_color = '#f8f9fa'
fig.patch.set_facecolor(bg_color)
ax1.set_facecolor(bg_color)
ax2.set_facecolor(bg_color)

def draw_vortex_structures(ax, x_positions, y_positions, sizes, colors, densities, title):
    """Draw vortex structures with varying complexity"""
    ax.set_xlim(-0.5, 10.5)
    ax.set_ylim(-0.5, 8.5)
    ax.set_aspect('equal')
    
    # Draw vortex spirals
    for i, (x, y, size, color, density) in enumerate(zip(x_positions, y_positions, sizes, colors, densities)):
        # Create spiral vortex pattern
        t = np.linspace(0, density * 2 * np.pi, int(100 * density))
        r = size * (1 - t / (density * 2 * np.pi))
        spiral_x = x + r * np.cos(t)
        spiral_y = y + r * np.sin(t)
        
        # Plot the spiral with varying line width
        for j in range(len(spiral_x)-1):
            line_width = 3 * (1 - j/(len(spiral_x)-1)) + 1
            ax.plot(spiral_x[j:j+2], spiral_y[j:j+2], color=color, 
                   linewidth=line_width, alpha=0.8)
        
        # Add rotation arrow
        arrow_radius = size * 0.7
        ax.annotate('', xy=(x + arrow_radius*0.7, y + arrow_radius*0.7),
                   xytext=(x + arrow_radius, y),
                   arrowprops=dict(arrowstyle='->', lw=2, color=color, alpha=0.9))
    
    ax.set_title(title, fontsize=16, fontweight='bold', pad=20)
    ax.set_xticks([])
    ax.set_yticks([])

# Standard Kolmogorov Turbulence (left panel)
# Dense, complex vortex structures
np.random.seed(42)
x_pos_k = np.random.uniform(1, 9, 20)
y_pos_k = np.random.uniform(1, 7, 20)
sizes_k = np.random.uniform(0.3, 1.0, 20)
colors_k = [cmap(0.8)] * 20
densities_k = np.random.uniform(2, 4, 20)

draw_vortex_structures(ax1, x_pos_k, y_pos_k, sizes_k, colors_k, densities_k, 
                      'Standard Kolmogorov Turbulence')

# Add energy injection arrow
ax1.annotate('Energy Injection', xy=(5, 7.5), xytext=(5, 8.2),
            arrowprops=dict(arrowstyle='->', lw=3, color='red'),
            ha='center', fontsize=12, fontweight='bold')

# Add energy cascade arrows
for i in range(3):
    x = 2 + i * 3
    ax1.annotate('', xy=(x+1.5, 4), xytext=(x, 4),
                arrowprops=dict(arrowstyle='->', lw=2, color='blue', alpha=0.7))

ax1.text(5, 0.2, r'$E_k \sim k^{-5/3}$ (Kolmogorov scaling)', 
         ha='center', fontsize=12, fontweight='bold',
         bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))

ax1.text(5, 4.5, r'Lamb force: $\vec{u} \times \vec{\omega}$', 
         ha='center', fontsize=11, style='italic',
         bbox=dict(boxstyle="round,pad=0.3", facecolor='lightblue', alpha=0.7))

# Gravitomagnetic Turbulence (right panel)
# Larger, more coherent structures with void regions
np.random.seed(24)
x_pos_g = [2, 7.5, 4, 8]
y_pos_g = [6, 5, 2.5, 1.5]
sizes_g = [1.2, 1.0, 1.3, 0.8]
colors_g = [cmap(0.3), cmap(0.5), cmap(0.4), cmap(0.6)]
densities_g = [1.5, 1.2, 1.3, 1.0]

draw_vortex_structures(ax2, x_pos_g, y_pos_g, sizes_g, colors_g, densities_g, 
                      'Gravitomagnetic Turbulence')

# Add large void regions
void1 = Circle((1.5, 4), 0.8, color='white', alpha=0.9, zorder=10)
void2 = Circle((6, 3), 1.0, color='white', alpha=0.9, zorder=10)
ax2.add_patch(void1)
ax2.add_patch(void2)
ax2.text(1.5, 4, 'VOID', ha='center', va='center', fontsize=10, 
         fontweight='bold', alpha=0.6)
ax2.text(6, 3, 'VOID', ha='center', va='center', fontsize=10, 
         fontweight='bold', alpha=0.6)

# Add energy injection arrow
ax2.annotate('Energy Injection', xy=(5, 7.5), xytext=(5, 8.2),
            arrowprops=dict(arrowstyle='->', lw=3, color='red'),
            ha='center', fontsize=12, fontweight='bold')

# Add gravitomagnetic field lines
t = np.linspace(0, 2*np.pi, 100)
for radius in [3, 4]:
    field_x = 5 + radius * np.cos(t)
    field_y = 4 + 0.5 * radius * np.sin(t)
    ax2.plot(field_x, field_y, '--', color='purple', alpha=0.6, linewidth=2)

ax2.text(5, 0.2, r'$E_k \sim k^{-3/2}$ (GMHD scaling)', 
         ha='center', fontsize=12, fontweight='bold',
         bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))

ax2.text(5, 4.8, r'$\vec{B}^g \times \vec{J}^g$ (gravitomagnetic force)', 
         ha='center', fontsize=11, style='italic',
         bbox=dict(boxstyle="round,pad=0.3", facecolor='lightyellow', alpha=0.7))

# Add legend
legend_elements = [
    mpatches.Patch(color=cmap(0.8), label='High vorticity regions'),
    mpatches.Patch(color=cmap(0.4), label='Coherent structures'),
    mpatches.Patch(color='white', label='Void regions', alpha=0.9),
    plt.Line2D([0], [0], color='purple', linestyle='--', label='Gravitomagnetic field', alpha=0.6)
]

fig.legend(handles=legend_elements, loc='lower center', ncol=4, 
          bbox_to_anchor=(0.5, 0.02), fontsize=11)

# Add parameter annotations
textstr = r'$N_g = \frac{|L_{\vec{B}^g} \vec{J}^g|}{|L_{\vec{u}} \vec{\omega}|}$ (gravitational interaction parameter)'
fig.text(0.5, 0.08, textstr, ha='center', fontsize=11, style='italic',
         bbox=dict(boxstyle="round,pad=0.5", facecolor='lightgray', alpha=0.8))

plt.tight_layout()
plt.subplots_adjust(bottom=0.15, top=0.90)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import cmocean
from matplotlib.patches import FancyBboxPatch
from matplotlib.patches import Circle
import matplotlib.patches as mpatches

# Set up the figure
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 10))
# fig.suptitle('Turbulent Energy Cascade: Kolmogorov vs Gravitomagnetic', 
#              fontsize=20, fontweight='bold', y=0.95)

# Color scheme using cmocean thermal
cmap = cmocean.cm.thermal
colors = [cmap(0.2), cmap(0.4), cmap(0.6), cmap(0.8)]
bg_color = '#f8f9fa'
fig.patch.set_facecolor(bg_color)
ax1.set_facecolor(bg_color)
ax2.set_facecolor(bg_color)

def create_turbulent_background(ax, density='high', alpha=0.25):
    """Create a swirly turbulent background pattern using thermal colormap"""
    x = np.linspace(-0.5, 10.5, 150)
    y = np.linspace(-0.5, 8.5, 120)
    X, Y = np.meshgrid(x, y)
    
    if density == 'high':
        # Scale-invariant Kolmogorov turbulence with swirly patterns
        # Multiple scales of vortical structures
        Z = (np.sin(X + 2*np.sin(Y)) * np.cos(Y + 2*np.cos(X)) + 
             0.7 * np.sin(2*X + 3*np.sin(1.5*Y)) * np.cos(2*Y + 3*np.cos(1.5*X)) +
             0.5 * np.sin(4*X + 4*np.sin(0.8*Y)) * np.cos(4*Y + 4*np.cos(0.8*X)) +
             0.3 * np.sin(6*X + 2*np.sin(2.5*Y)) * np.cos(6*Y + 2*np.cos(2.5*X)) +
             0.2 * np.sin(10*X + 1.5*np.sin(3*Y)) * np.cos(10*Y + 1.5*np.cos(3*X)))
    else:
        # Gravitomagnetic turbulence with large structures and voids
        # Create swirly large-scale patterns with distinct void regions
        base_pattern = np.sin(0.8*X + 1.5*np.sin(0.6*Y)) * np.cos(0.8*Y + 1.5*np.cos(0.6*X))
        medium_pattern = 0.4 * np.sin(1.5*X + 2*np.sin(0.4*Y)) * np.cos(1.5*Y + 2*np.cos(0.4*X))
        
        # Create void regions by suppressing activity in certain areas
        void_mask1 = np.exp(-((X-2)**2 + (Y-4)**2)/3)
        void_mask2 = np.exp(-((X-6.5)**2 + (Y-3)**2)/4) 
        void_mask3 = np.exp(-((X-8)**2 + (Y-6)**2)/2.5)
        
        Z = base_pattern + medium_pattern
        Z = Z * (1 - 0.8*void_mask1) * (1 - 0.7*void_mask2) * (1 - 0.6*void_mask3)
    
    # Normalize and apply thermal colormap
    Z_norm = (Z - Z.min()) / (Z.max() - Z.min())
    ax.imshow(Z_norm, extent=[-0.5, 10.5, -0.5, 8.5], cmap=cmap, alpha=alpha, aspect='auto')

def draw_vortex_structures(ax, x_positions, y_positions, sizes, colors, densities, title, bg_type='high'):
    """Draw vortex structures with varying complexity"""
    ax.set_xlim(-0.5, 10.5)
    ax.set_ylim(-0.5, 8.5)
    ax.set_aspect('equal')
    
    # Add turbulent background
    create_turbulent_background(ax, density=bg_type, alpha=0.25)
    
    # Draw vortex spirals
    for i, (x, y, size, color, density) in enumerate(zip(x_positions, y_positions, sizes, colors, densities)):
        # Create spiral vortex pattern
        t = np.linspace(0, density * 2 * np.pi, int(100 * density))
        r = size * (1 - t / (density * 2 * np.pi))
        spiral_x = x + r * np.cos(t)
        spiral_y = y + r * np.sin(t)
        
        # Plot the spiral with varying line width
        for j in range(len(spiral_x)-1):
            line_width = 3 * (1 - j/(len(spiral_x)-1)) + 1
            ax.plot(spiral_x[j:j+2], spiral_y[j:j+2], color=color, 
                   linewidth=line_width, alpha=0.9)
        
        # Add rotation arrow
        arrow_radius = size * 0.7
        ax.annotate('', xy=(x + arrow_radius*0.7, y + arrow_radius*0.7),
                   xytext=(x + arrow_radius, y),
                   arrowprops=dict(arrowstyle='->', lw=2, color=color, alpha=0.9))
    
    ax.set_title(title, fontsize=16, fontweight='bold', pad=20)
    ax.set_xticks([])
    ax.set_yticks([])

# Standard Kolmogorov Turbulence (left panel)
# Dense, complex vortex structures with more variety
np.random.seed(42)

# Large eddies
x_large_k = np.random.uniform(1, 9, 8)
y_large_k = np.random.uniform(1.5, 6.5, 8)
sizes_large_k = np.random.uniform(0.8, 1.4, 8)
colors_large_k = [cmap(0.9)] * 8
densities_large_k = np.random.uniform(2, 3.5, 8)

# Medium eddies
x_med_k = np.random.uniform(0.5, 9.5, 15)
y_med_k = np.random.uniform(1, 7, 15)
sizes_med_k = np.random.uniform(0.4, 0.8, 15)
colors_med_k = [cmap(0.7)] * 15
densities_med_k = np.random.uniform(2.5, 4, 15)

# Small eddies
x_small_k = np.random.uniform(0.2, 9.8, 25)
y_small_k = np.random.uniform(0.8, 7.2, 25)
sizes_small_k = np.random.uniform(0.15, 0.4, 25)
colors_small_k = [cmap(0.5)] * 25
densities_small_k = np.random.uniform(3, 5, 25)

# Combine all eddies
x_pos_k = np.concatenate([x_large_k, x_med_k, x_small_k])
y_pos_k = np.concatenate([y_large_k, y_med_k, y_small_k])
sizes_k = np.concatenate([sizes_large_k, sizes_med_k, sizes_small_k])
colors_k = colors_large_k + colors_med_k + colors_small_k
densities_k = np.concatenate([densities_large_k, densities_med_k, densities_small_k])

draw_vortex_structures(ax1, x_pos_k, y_pos_k, sizes_k, colors_k, densities_k, 
                      'Standard Kolmogorov Turbulence', bg_type='high')

# Add energy injection arrow
ax1.annotate('Energy Injection', xy=(5, 7.5), xytext=(5, 8.2),
            arrowprops=dict(arrowstyle='->', lw=3, color='red'),
            ha='center', fontsize=12, fontweight='bold')

# Add energy cascade arrows
# for i in range(3):
#     x = 2 + i * 3
#     ax1.annotate('', xy=(x+1.5, 4), xytext=(x, 4),
#                 arrowprops=dict(arrowstyle='->', lw=2, color='blue', alpha=0.7))

ax1.text(5, 0.2, r'$E_k \sim k^{-5/3}$ (Kolmogorov scaling)', 
         ha='center', fontsize=15, fontweight='bold',
         bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))

ax1.text(5, 4.5, '', 
         ha='center', fontsize=11, style='italic',
         bbox=dict(boxstyle="round,pad=0.3", facecolor='lightblue', alpha=0.7))

# Gravitomagnetic Turbulence (right panel)
# Larger, more coherent structures with void regions and NO overlapping
np.random.seed(24)

# Large coherent structures positioned to avoid void regions
x_large_g = [1.5, 8.5, 4.5, 9.2, 0.8]
y_large_g = [6.5, 1.8, 0.8, 5.5, 2]
sizes_large_g = [1.0, 0.9, 1.2, 0.8, 1.1]
colors_large_g = [cmap(0.4), cmap(0.5), cmap(0.3), cmap(0.6), cmap(0.4)]
densities_large_g = [1.2, 1.0, 1.3, 1.1, 0.9]

# Medium structures carefully positioned
x_med_g = [3.2, 7, 1, 5.8, 9.5]
y_med_g = [7, 7.2, 0.8, 6.5, 3.8]
sizes_med_g = [0.5, 0.6, 0.4, 0.5, 0.4]
colors_med_g = [cmap(0.3), cmap(0.4), cmap(0.5), cmap(0.4), cmap(0.3)]
densities_med_g = [1.5, 1.3, 1.4, 1.2, 1.6]

# Very few small structures (suppressed by gravitomagnetic effects)
x_small_g = [6.5, 3.8, 0.3]
y_small_g = [0.5, 1.5, 5.5]
sizes_small_g = [0.25, 0.3, 0.2]
colors_small_g = [cmap(0.2), cmap(0.25), cmap(0.3)]
densities_small_g = [2, 1.8, 2.2]

# Combine all eddies
x_pos_g = x_large_g + x_med_g + x_small_g
y_pos_g = y_large_g + y_med_g + y_small_g
sizes_g = sizes_large_g + sizes_med_g + sizes_small_g
colors_g = colors_large_g + colors_med_g + colors_small_g
densities_g = densities_large_g + densities_med_g + densities_small_g

draw_vortex_structures(ax2, x_pos_g, y_pos_g, sizes_g, colors_g, densities_g, 
                      'Gravitomagnetic Turbulence', bg_type='low')

# Add large void regions that don't overlap with eddies
void1 = Circle((2, 4), 1.2, color='white', alpha=0.85, zorder=5)
void2 = Circle((6.5, 3), 1.4, color='white', alpha=0.85, zorder=5) 
void3 = Circle((8, 6), 0.9, color='white', alpha=0.8, zorder=5)
ax2.add_patch(void1)
ax2.add_patch(void2)
ax2.add_patch(void3)
ax2.text(2, 4, 'VOID', ha='center', va='center', fontsize=11, 
         fontweight='bold', alpha=0.7)
ax2.text(6.5, 3, 'VOID', ha='center', va='center', fontsize=11, 
         fontweight='bold', alpha=0.7)
ax2.text(8, 6, 'VOID', ha='center', va='center', fontsize=10, 
         fontweight='bold', alpha=0.7)

# Add energy injection arrow
ax2.annotate('Energy Injection', xy=(5, 7.5), xytext=(5, 8.2),
            arrowprops=dict(arrowstyle='->', lw=3, color='red'),
            ha='center', fontsize=12, fontweight='bold')

# Add gravitomagnetic field lines
t = np.linspace(0, 2*np.pi, 100)
for radius in [3, 4]:
    field_x = 5 + radius * np.cos(t)
    field_y = 4 + 0.5 * radius * np.sin(t)
    ax2.plot(field_x, field_y, '--', color='purple', alpha=0.6, linewidth=2)

ax2.text(5, 0.2, r'$E_k \sim k^{-3/2}$ (GMHD scaling)', 
         ha='center', fontsize=15, fontweight='bold',
         bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))

ax2.text(5, 4.8, '', 
         ha='center', fontsize=11, style='italic',
         bbox=dict(boxstyle="round,pad=0.3", facecolor='lightyellow', alpha=0.7))

# Add legend
legend_elements = [
    mpatches.Patch(color=cmap(0.8), label='High vorticity regions'),
    mpatches.Patch(color=cmap(0.4), label='Coherent structures'),
    mpatches.Patch(color='white', label='Void regions', alpha=0.9),
    plt.Line2D([0], [0], color='purple', linestyle='--', label='Gravitomagnetic field', alpha=0.6)
]

fig.legend(handles=legend_elements, loc='lower center', ncol=4, 
          bbox_to_anchor=(0.5, 0.02), fontsize=11)

# Add parameter annotations
# textstr = r'$N_g = \frac{|L_{\vec{B}^g} \vec{J}^g|}{|L_{\vec{u}} \vec{\omega}|}$ (gravitational interaction parameter)'
# fig.text(0.5, 0.08, textstr, ha='center', fontsize=11, style='italic',
#          bbox=dict(boxstyle="round,pad=0.5", facecolor='lightgray', alpha=0.8))

plt.tight_layout()
plt.subplots_adjust(bottom=0.15, top=0.90)
plt.show()
```
