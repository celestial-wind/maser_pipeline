```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.patches import Circle, FancyArrowPatch
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
import matplotlib.patches as mpatches
from matplotlib.colors import LinearSegmentedColormap
import cmocean

# Set up the figure with high DPI for publication quality
fig = plt.figure(figsize=(16, 12), dpi=300)
ax = fig.add_subplot(111, projection='3d')

# Color schemes for different components
jet_colors = cmocean.cm.thermal
disk_colors = cmocean.cm.thermal
field_colors = cmocean.cm.deep

# Black hole parameters
M = 1.0  # Mass in arbitrary units
a = 0.9  # Spin parameter (near-extremal)
r_s = 2 * M  # Schwarzschild radius
r_erg = M + np.sqrt(M**2 - a**2)  # Ergosphere outer boundary

# Create coordinate grids
theta = np.linspace(0, 2*np.pi, 100)
phi = np.linspace(0, np.pi, 50)

# 1. BLACK HOLE EVENT HORIZON
r_horizon = M + np.sqrt(M**2 - a**2)
u = np.linspace(0, 2 * np.pi, 50)
v = np.linspace(0, np.pi, 50)
x_bh = r_horizon * np.outer(np.cos(u), np.sin(v))
y_bh = r_horizon * np.outer(np.sin(u), np.sin(v))
z_bh = r_horizon * np.outer(np.ones(np.size(u)), np.cos(v))
ax.plot_surface(x_bh, y_bh, z_bh, color='black', alpha=0.9, linewidth=0)

# 2. ERGOSPHERE (gravitomagnetic "battery" region)
r_erg_surface = M + np.sqrt(M**2 - a**2 * np.cos(phi)**2)
for i in range(0, len(phi), 5):
    x_erg = r_erg_surface[i] * np.cos(theta) * np.sin(phi[i])
    y_erg = r_erg_surface[i] * np.sin(theta) * np.sin(phi[i])
    z_erg = r_erg_surface[i] * np.cos(phi[i]) * np.ones_like(theta)
    ax.plot(x_erg, y_erg, z_erg, color='gold', alpha=0.3, linewidth=1)

# 3. WARPED ACCRETION DISK with Lense-Thirring precession
r_disk = np.linspace(3*r_s, 15*r_s, 30)
theta_disk = np.linspace(0, 2*np.pi, 100)
R, T = np.meshgrid(r_disk, theta_disk)

# Disk warping due to frame-dragging (Lense-Thirring effect)
warp_amplitude = 0.3 * r_s
precession_rate = 2 * a * M / R**3  # Simplified precession
Z_warp = warp_amplitude * np.sin(T - precession_rate * 10) * np.exp(-R/(8*r_s))

X_disk = R * np.cos(T)
Y_disk = R * np.sin(T)
Z_disk = Z_warp

# Color the disk by temperature (closer = hotter)
temp_colors = 1 / (R + 1)  # Simplified temperature profile
disk_surface = ax.plot_surface(X_disk, Y_disk, Z_disk, 
                              facecolors=disk_colors(temp_colors),
                              alpha=0.7, linewidth=0, antialiased=True)

# 4. RELATIVISTIC JETS with collimation
jet_height = np.linspace(0, 25*r_s, 50)
jet_opening_angle = 5 * np.pi/180  # 5 degree opening angle

# Northern jet
jet_radius_n = np.abs(jet_height - 5*r_s) * np.tan(jet_opening_angle) + 0.5*r_s
jet_radius_n = np.maximum(jet_radius_n, 0.3*r_s)  # Minimum radius for collimation

# Jet precession due to frame-dragging
precession_angle = 0.1 * np.sin(jet_height / (2*r_s))

for i in range(0, len(jet_height), 3):
    if jet_height[i] > 2*r_s:  # Start jets above the disk
        theta_jet = np.linspace(0, 2*np.pi, 20)
        x_jet = jet_radius_n[i] * np.cos(theta_jet + precession_angle[i])
        y_jet = jet_radius_n[i] * np.sin(theta_jet + precession_angle[i])
        z_jet = jet_height[i] * np.ones_like(theta_jet)
        
        # Color by velocity (blue-shifted for approaching jet)
        jet_color = jet_colors(0.8 - 0.4 * i/len(jet_height))
        ax.plot(x_jet, y_jet, z_jet, color=jet_color, alpha=0.8, linewidth=2)

# Southern jet (red-shifted, receding)
for i in range(0, len(jet_height), 3):
    if jet_height[i] > 2*r_s:
        theta_jet = np.linspace(0, 2*np.pi, 20)
        x_jet = jet_radius_n[i] * np.cos(theta_jet - precession_angle[i])
        y_jet = jet_radius_n[i] * np.sin(theta_jet - precession_angle[i])
        z_jet = -jet_height[i] * np.ones_like(theta_jet)
        
        jet_color = jet_colors(0.2 + 0.4 * i/len(jet_height))
        ax.plot(x_jet, y_jet, z_jet, color=jet_color, alpha=0.8, linewidth=2)

# 5. MAGNETIC FIELD LINES showing energy extraction
r_field = np.linspace(r_horizon*1.1, 20*r_s, 20)
for i in range(0, 360, 30):
    phi_field = np.radians(i)
    
    # Field lines threading the ergosphere (energy extraction)
    x_field = r_field * np.cos(phi_field)
    y_field = r_field * np.sin(phi_field)
    z_field = 2*r_s * np.sin(r_field / (5*r_s)) * np.cos(phi_field + a*0.1)
    
    ax.plot(x_field, y_field, z_field, color='cyan', alpha=0.6, linewidth=1.5)
    ax.plot(x_field, y_field, -z_field, color='cyan', alpha=0.6, linewidth=1.5)

# 6. FRAME-DRAGGING VECTORS
drag_radius = np.array([3*r_s, 5*r_s, 8*r_s, 12*r_s])
for r in drag_radius:
    for angle in np.linspace(0, 2*np.pi, 8):
        x_start = r * np.cos(angle)
        y_start = r * np.sin(angle)
        z_start = 0
        
        # Dragging velocity proportional to a*M/r³
        drag_velocity = a * M / r**3
        dx = -drag_velocity * r * np.sin(angle) * 3
        dy = drag_velocity * r * np.cos(angle) * 3
        
        ax.quiver(x_start, y_start, z_start, dx, dy, 0,
                 color='yellow', alpha=0.7, arrow_length_ratio=0.1,
                 linewidth=2)

# 7. ENERGY EXTRACTION ARROWS (Penrose mechanism)
extraction_points_r = [4*r_s, 6*r_s, 8*r_s]
extraction_points_theta = [np.pi/4, np.pi/2, 3*np.pi/4, 5*np.pi/4]

for r in extraction_points_r:
    for theta_ext in extraction_points_theta:
        x_ext = r * np.cos(theta_ext)
        y_ext = r * np.sin(theta_ext)
        z_ext = 0
        
        # Energy flow direction
        energy_flow_r = 2*r_s
        energy_flow_x = energy_flow_r * np.cos(theta_ext + np.pi/2)
        energy_flow_y = energy_flow_r * np.sin(theta_ext + np.pi/2)
        energy_flow_z = 3*r_s
        
        ax.quiver(x_ext, y_ext, z_ext, 
                 energy_flow_x, energy_flow_y, energy_flow_z,
                 color='orange', alpha=0.8, arrow_length_ratio=0.1,
                 linewidth=3)

# ANNOTATIONS AND LABELS
ax.text(0, 0, 30*r_s, 'Relativistic Jet\n(Doppler boosted)', 
        fontsize=12, ha='center', weight='bold', color='white',
        bbox=dict(boxstyle="round,pad=0.3", facecolor='black', alpha=0.7))

ax.text(15*r_s, 0, 0, 'Warped Accretion Disk\n(Lense-Thirring precession)', 
        fontsize=11, ha='center', weight='bold', color='red',
        bbox=dict(boxstyle="round,pad=0.3", facecolor='white', alpha=0.8))

ax.text(-8*r_s, -8*r_s, 5*r_s, 'Frame-dragging\n(Gravitomagnetism)', 
        fontsize=11, ha='center', weight='bold', color='yellow',
        bbox=dict(boxstyle="round,pad=0.3", facecolor='black', alpha=0.7))

ax.text(2*r_s, 8*r_s, 8*r_s, 'Energy Extraction\n(Penrose Mechanism)', 
        fontsize=11, ha='center', weight='bold', color='orange',
        bbox=dict(boxstyle="round,pad=0.3", facecolor='black', alpha=0.7))

ax.text(4*r_s, 0, -5*r_s, 'Ergosphere\n("Gravitomagnetic Battery")', 
        fontsize=11, ha='center', weight='bold', color='gold',
        bbox=dict(boxstyle="round,pad=0.3", facecolor='black', alpha=0.7))

ax.text(-12*r_s, 12*r_s, 0, 'Magnetic Field Lines\n(Threading Ergosphere)', 
        fontsize=10, ha='center', weight='bold', color='cyan',
        bbox=dict(boxstyle="round,pad=0.3", facecolor='black', alpha=0.7))

# Create legend
legend_elements = [
    mpatches.Patch(color='black', label='Event Horizon'),
    mpatches.Patch(color='gold', alpha=0.5, label='Ergosphere'),
    mpatches.Patch(color='red', label='Accretion Disk'),
    mpatches.Patch(color='purple', label='Relativistic Jets'),
    mpatches.Patch(color='cyan', label='Magnetic Field'),
    mpatches.Patch(color='yellow', label='Frame-dragging'),
    mpatches.Patch(color='orange', label='Energy Extraction')
]

ax.legend(handles=legend_elements, loc='upper left', bbox_to_anchor=(0, 1),
          fontsize=10, framealpha=0.9)

# Set axis properties
ax.set_xlim([-20*r_s, 20*r_s])
ax.set_ylim([-20*r_s, 20*r_s])
ax.set_zlim([-20*r_s, 30*r_s])

ax.set_xlabel('X (Schwarzschild radii)', fontsize=12, weight='bold')
ax.set_ylabel('Y (Schwarzschild radii)', fontsize=12, weight='bold')
ax.set_zlabel('Z (Schwarzschild radii)', fontsize=12, weight='bold')

# Set viewing angle for best visualization
ax.view_init(elev=20, azim=45)

# Title and subtitle
plt.suptitle('Gravitomagnetic Effects in Black Hole Jets', 
             fontsize=18, weight='bold', y=0.95)
plt.title('3D Visualization of Energy Extraction, Frame-dragging, and Jet Collimation\n' +
          f'Kerr Black Hole (a/M = {a}, M = {M})', 
          fontsize=12, pad=20)

# Add publication info
fig.text(0.02, 0.02, 
         'Key Features:\n'
         '• Ergosphere as gravitomagnetic "battery"\n'
         '• Lense-Thirring disk warping & precession\n'
         '• Magnetic field threading & energy extraction\n'
         '• Jet collimation & frame-dragging effects\n'
         '• Penrose mechanism energy extraction',
         fontsize=9, verticalalignment='bottom',
         bbox=dict(boxstyle="round,pad=0.5", facecolor='lightgray', alpha=0.8))

# Remove axis tick marks for cleaner look
ax.set_xticks([])
ax.set_yticks([])
ax.set_zticks([])

# Set background color
ax.xaxis.pane.fill = False
ax.yaxis.pane.fill = False
ax.zaxis.pane.fill = False
ax.xaxis.pane.set_edgecolor('white')
ax.yaxis.pane.set_edgecolor('white')
ax.zaxis.pane.set_edgecolor('white')
ax.xaxis.pane.set_alpha(0.1)
ax.yaxis.pane.set_alpha(0.1)
ax.zaxis.pane.set_alpha(0.1)

plt.tight_layout()
plt.show()

# Print technical details for publication
print("TECHNICAL DETAILS FOR PUBLICATION:")
print("="*50)
print(f"Black Hole Parameters:")
print(f"  - Mass: {M} (normalized units)")
print(f"  - Spin parameter (a/M): {a}")
print(f"  - Schwarzschild radius: {r_s:.1f}")
print(f"  - Ergosphere radius: {r_erg:.2f}")
print(f"\nKey Gravitomagnetic Effects Visualized:")
print(f"  - Frame-dragging frequency: ∝ a·M/r³")
print(f"  - Lense-Thirring precession of disk")
print(f"  - Energy extraction via Penrose mechanism")
print(f"  - Jet collimation via magnetic fields")
print(f"  - Ergosphere as 'gravitomagnetic battery'")
print(f"\nVisualization includes:")
print(f"  - 3D warped spacetime effects")
print(f"  - Magnetic field line threading")
print(f"  - Doppler-shifted jet emission")
print(f"  - Temperature-coded accretion disk")
print(f"  - Frame-dragging velocity vectors")
```
